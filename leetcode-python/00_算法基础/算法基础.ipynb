# 3.递归算法的时间复杂度

例题：求x^n的时间复杂度



```python
# n次递归,复杂度O（n）
# 递归的次数n*每次递归的操作次数1
def mypoly(x,n):
    if n == 0 : return 1 # 终止条件
    return mypoly(x,n-1)*x

mypoly(4,3)
```

```python
# 递归深度O(logn)，但每层计算次数翻倍，总时间复杂度仍为O(n)
def mypoly2(x,n):
    if n == 0 : return 1 # 偶数的终止条件
    if n == 1 : return x # 奇数的终止条件
    if n % 2 == 1:
        return mypoly2(x,int(n/2)) * mypoly2(x,int(n/2)) * x
    return mypoly2(x,int(n/2)) * mypoly2(x,int(n/2))

mypoly2(4,3)
```

```python
# 真正的O(logn)
def mypoly3(x,n):
    if n==0: return 1
    if n==1: return x
    t = mypoly3(x, n//2)
    if n % 2 == 1: return t *t *x
    return t*t
mypoly3(4,3)
```

# 5.递归算法的时空复杂度
#### 斐波那契数列

```python
# 原始
def fibo(i):
    if i<=0: return 0
    if i == 1: return 1
    return fibo(i-1)+fibo(i-2)

fibo(8)
```

- 时间复杂度：n^2,随n增大指数上升

```python
# 优化后的递归算法（减少递归调用次数）
def fibo1(first, second, n):
    if n<=0: return 0
    if n < 3: return 1
    elif n == 3: return first + second
    else: return fibo1(second, first + second, n-1)
# 时间和空间复杂度都是O(n)
```
