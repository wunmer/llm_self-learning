## 001.赋值，浅拷贝，深拷贝

````python
lst = [1,2,3,4]
list2 = lst # 直接赋值
lst.append(5) # 给lst新增元素：直接修改变量名lst所代表的列表对象，没有创建新元素
print(lst)
print(list2)
# 打印一致
print(id(lst))
print(id(list2))
# 内存地址一致


lst=[1,2,3] # 给lst重新赋值：没有修改原来的[1,2,3,4]对象，而是新建了一个列表对象[1,2,3]
# 让lst变量名从列表[1,2,3,4]上撕下来，贴到了新的[1,2,3]列表上
print(lst)
print(list2)
# 打印不一致
print(id(lst))
print(id(list2))
# 内存地址不一致
````

````python
# 浅拷贝:只复制外层，内层可变对象不复制
a = [1,2,[3,4]]
b = a.copy()
b.append(5)
print(a)
print(b)
b[2].append(99)
print(a)
print(b)
print(id(a))
print(id(b))
# 内存地址不一致
print(id(a[2]))
print(id(b[2]))
# 列表内部元素内存地址一致
````

````python
import copy
# 深拷贝:递归复制所有层，内外都是全新对象，彻底独立
a = [1,2,[3,4]]
b = copy.deepcopy(a)
b.append(5)
print(a)
print(b)
b[2].append(99)
print(a)
print(b)
print(id(a))
print(id(b))
print(id(a[2]))
print(id(b[2]))
# 列表地址和内部元素地址都不一致
````

````python
# tips:不能用list这种内置类型做变量名。如果已经错误地给list列表复制了，则用如下方法删除变量：
list = [1,2,3]
del list
# 现在变量名删除，list类型回到了它原本的用法
print(list([1,2,3]))
````

## 002.可变对象

- 存储空间保存的数据允许被修改，但内存不会发生改变，就是可变类型
- 常见的可变对象类型：
    - 列表list
    - 字典dict
    - 集合set
- 不可变对象：只能赋值一个新对象，赋值后内存会发生改变

## 003.匿名函数（lambda函数）

````python
# 普通函数
def add(a,b):
    return a+b
print(add(1,3))

# 匿名函数
add = lambda a,b:a+b
print(add(1,3))
````

````python
# lambda的不同参数形式：
# 1.无参数
drink = lambda :"水果茶"
print(drink())

# 2.一个参数
drink = lambda a:a
print(drink("水果茶"))

# 3.默认参数(注意：返回值应当以元组的形式)
info = lambda name,age=18:(name,age)
print(info("水果茶"))

# 4.关键字参数
fund = lambda **kwargs:kwargs
print(fund(name = 'wr',age = '21'))
````

## 004.关键字参数和位置参数的打包传入

- 什么是```**kwargs```: 
    - ```**kwargs```是python里接收任意数量关键字参数的写法。kwargs是keyword arguments的缩写，```**abc```，```**options```都可表达相同意思。
    - 前面的```**```表示把调用时传入的所有关键字参数，打包为一个字典
- 什么是关键字参数：
    - “关键字参数”指调用函数时用```参数名=值```的形式传参

````python
def f(a,b):
    print(a,b)
f(1,2) # 位置参数：按顺序对应
f(b=2,a=1) # 关键字参数：按参数名对应，顺序可以打乱
````

- 位置参数：```*args```（注意```**kwargs```和```*args```与C++的指针语法不是一个东西）

````python
f = lambda *args,**kargs:(args, kargs)
print(f(1,2,3,5,name = 'wr', age = 21))
````
