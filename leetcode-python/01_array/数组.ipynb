# 2.二分查找
https://leetcode.cn/problems/binary-search/

二分法：对于从大到小或者从小到大排列的（有序、无重复元素）数组，每一轮使用mid与查找目标比较，最终查找时间复杂度为log(n)

- 二分法第一种写法（左闭右闭）：
    - while的终止条件要用left<=right
    - nums[middle] < target判断后，left要赋值为mid+1，因为当前middle一定不是target。若赋值为mid可能导致一直无法跳出
    - 同理right赋值mid-1

- 二分法第二种写法（左闭右开）：
    - while的终止条件left < right
    - 与第一种写法的区别：nums[middle]>target后，right赋值为mid

# 5.长度最小的子数组
https://leetcode.cn/problems/minimum-size-subarray-sum/

给定一个含有 n 个正整数的数组和一个正整数 target 。

找出该数组中满足其总和大于等于 target 的长度最小的 子数组 [numsl, numsl+1, ..., numsr-1, numsr] ，并返回其长度。如果不存在符合条件的子数组，返回 0 。（注：“子数组”意为来自原数组中相邻的元素）

示例：

输入：s = 7, nums = [2,3,1,2,4,3]

输出：2

解释：子数组 [4,3] 是该条件下的长度最小的子数组。
