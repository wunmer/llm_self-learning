# 2.二分查找
https://leetcode.cn/problems/binary-search/

二分法：对于从大到小或者从小到大排列的（有序、无重复元素）数组，每一轮使用mid与查找目标比较，最终查找时间复杂度为log(n)

- 二分法第一种写法（左闭右闭）：
    - while的终止条件要用left<=right
    - nums[middle] < target判断后，left要赋值为mid+1，因为当前middle一定不是target。若赋值为mid可能导致一直无法跳出
    - 同理right赋值mid-1

- 二分法第二种写法（左闭右开）：
    - while的终止条件left < right
    - 与第一种写法的区别：nums[middle]>target后，right赋值为mid

# 5.长度最小的子数组
https://leetcode.cn/problems/minimum-size-subarray-sum/

给定一个含有 n 个正整数的数组和一个正整数 target 。

找出该数组中满足其总和大于等于 target 的长度最小的 子数组 [numsl, numsl+1, ..., numsr-1, numsr] ，并返回其长度。如果不存在符合条件的子数组，返回 0 。（注：“子数组”意为来自原数组中相邻的元素）

示例：

输入：s = 7, nums = [2,3,1,2,4,3]

输出：2

解释：子数组 [4,3] 是该条件下的长度最小的子数组。

**暴力解：两层for循环**，时间复杂度O(n^2)

```python
# C++解：
# class Solution {
# public:
#     int minSubArrayLen(int s, vector<int>& nums) {
#         int result = INT32_MAX; // 记录遍历过程中每一步得到的最小子数组长度
#           //INT32_MAX是C++标准库中提供的一个宏，定义在头文件<cstdint>中，不是变量，没有类型，没有地址。
#         int sum = 0; // 子序列的数值之和
#         int subLength = 0; // 子序列的长度
#         for (int i = 0; i < nums.size(); i++) { // 设置子序列起点为i
#             sum = 0;
#             for (int j = i; j < nums.size(); j++) { // 设置子序列终止位置为j
#                 sum += nums[j];
#                 if (sum >= s) { // 一旦发现子序列和超过了s，更新result
#                     subLength = j - i + 1; // 取子序列的长度
#                     result = result < subLength ? result : subLength;
#                       //因为一旦找到某个i下的最小子数组，之后的j带来的数组都大于target且都比该数组长，所以break
#                     break; // 此处break是跳出内层for循环，直接进入到下一个i取值的for循环中
#                 }
#             }
#         }
#         // 如果result没有被赋值的话，就返回0，说明没有符合条件的子序列
#         return result == INT32_MAX ? 0 : result;
#     }
# };
```

```python
import time 
import random 

nums = [random.randint(1, 10) for _ in range(10000)]
target = 5000

# 模仿C++解法的python暴力解，时间复杂度O(n^2)：
def minsubarray(nums, target):
    result = 10000
    for i in range(len(nums)):
        sum = 0
        for j in range(i, len(nums)):
            sum += nums[j]
            sublength = j - i + 1
            if sum >= target:
                result = sublength if sublength <= result else result
                break
    return 0 if result == 10000 else result

begin=time.perf_counter()
minsubarray(nums, target)
end=time.perf_counter()
print(end-begin)
```

**滑动窗口**（双指针）

不断调节子序列的开始和结束位置，得出想要的结果
- 在暴力解法中，一个for循环标记窗口起始位置，另一个for循环标记窗口结束位置，用两个for循环完成搜索区间的过程
- 滑动窗口法用一个for循环完成这个任务：
    - 滑动窗口法把for循环安在窗口结束位置上
    - 每个元素进入滑动窗口被操作一次，出滑动窗口被操作一次，时间复杂度是2n即O(n^2)

```python
def minsubarray(nums, target):
    result = 10000
    i = 0 
    sum = 0
    for j in range(len(nums)):
        sum += nums[j]
        while sum >= target:
            sublength = j - i + 1
            result = sublength if sublength <= result else result
            sum -= nums[i]
            i += 1
    return result if result != 10000 else 0

begin=time.perf_counter()
minsubarray(nums, target)
end=time.perf_counter()
print(end-begin)
```
