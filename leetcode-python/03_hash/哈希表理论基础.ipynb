- 哈希表的作用：快速判断元素是否出现、统计频率和建立映射，高频用于数组和字符串题目
- 核心：根据键值范围，在数组、集合和映射之间选择合适的数据结构

- 哈希函数：有一个字符串列表，需要查找某个字符串是否存在于列表中
    - 哈希函数把每个字符串直接映射为数值，即把字符串映射为哈希表上的索引数字
- 哈希碰撞：哈希表索引长度短于字符串列表映射后的序列长度，两个字符串映射到了同一个索引
    - 解决方法：
        - 拉链法：发生冲突的元素都被存储在以相应索引为head的链表中
        - 线性探测法：冲突的位置已经放了一个字符串，就把冲突的另一个字符串放到下一个空位（强制要求哈希表大小tablesize大于字符串序列长度datasize）

**红黑树**

- 主流set和multiset的STL实现都使用红黑树
- 红黑树：一种自平衡二叉搜索树。它在普通二叉树的基础上，给每个节点增加一个颜色：红色或黑色，并通过一些规则保证红黑树不退化为链表
- 红黑树必须满足以下5条性质：
    - 每个节点只能是红色或者黑色
    - 根节点黑
    - 每个叶子节点是黑色
    - 不能连续出现两个红色节点
    - 从任意一个节点出发到叶子节点的路径上，黑色节点的数量相同
- 这些性质保证：
    - 最长路径不会超过最短路径的两倍
    - 树高保持在O(logn)，因此查找、插入、删除都是O(logn)
